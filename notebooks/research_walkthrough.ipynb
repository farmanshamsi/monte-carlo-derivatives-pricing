## Methods

The engine compares exact terminal GBM, Euler–Maruyama, and Milstein simulation against Black–Scholes benchmarks. Euler and Milstein use independent Brownian increments at every time step. Antithetic sampling and a discounted-stock control variate reduce estimator variance.

## Interpretation

The project is a numerical-methods validation study, not a claim that GBM is a complete market model.
